```python
# Dataset summary
import json
import pandas as pd

records = []
for path in OUTPUT_DIR.glob('**/*.json'):
    doc = json.loads(path.read_text())
    records.append({'category': doc['category'], 'words': doc['num_words'], 'doc_id': doc['doc_id']})

df = pd.DataFrame(records)
print(df.groupby('category').agg(count=('doc_id','count'), avg_words=('words','mean')).round(0))
```
