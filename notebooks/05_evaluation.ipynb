```python
import sys
sys.path.insert(0, '..')

from src.evaluate import evaluate, print_summary, correlate_with_human
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('../results/evaluation.csv')
df.head()
```

```python
# Heatmap: Intent-F1 by method and category
pivot = df.pivot_table(values='intent_f1', index='category', columns='method', aggfunc='mean')

fig, ax = plt.subplots(figsize=(8, 4))
sns.heatmap(pivot, annot=True, fmt='.3f', cmap='YlGnBu', ax=ax)
ax.set_title('Intent-F1 by Category and Method')
plt.tight_layout()
plt.savefig('../results/intent_f1_heatmap.png', dpi=150)
plt.show()
```

```python
# Bar chart: overall method comparison
metrics = ['intent_precision', 'intent_recall', 'intent_f1', 'rouge1_f', 'rougeL_f']
agg = df.groupby('method')[metrics].mean()

agg.plot(kind='bar', figsize=(10, 5), rot=0)
plt.title('Method Comparison — All Metrics')
plt.ylabel('Score')
plt.legend(loc='upper right')
plt.tight_layout()
plt.savefig('../results/method_comparison.png', dpi=150)
plt.show()

print(agg.round(4).to_string())
```
